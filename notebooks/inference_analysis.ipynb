```python

df = pd.read_json(project_root+"\\"+"inference_logs.jsonl", lines=True)


# перевод из мс в секунды
df["inference_time_sec"] = df["inference_time_ms"] / 1000

# на всякий случай
df["timestamp"] = pd.to_datetime(df["timestamp"])
```
